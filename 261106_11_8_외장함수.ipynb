```python
# 11.8.2 운영체제의 기본 기능 모듈
# os 모듈
# 운영체제에서 제공하는 기본 기능
# ex. 폴더 생성 및 삭제

# 현재 작업 홀더 위치(경로)
import os
print(os.getcwd())
```

```python
# exists(): 주어진 경로에 해당하는 폴더 또는 파일이 존재하는지 여부 확인
# makedirs(): 현재 위치에 폴더 새로 생성
# rmdir(): 현재 위치에 같은 이름의 폴더가 있으면 폴더 삭제
# listdir(): 현재 작업 폴더 안에 있는 폴더와 파일 목록 출력

import os

folder = "sample_dir"

if os.path.exists(folder): # 같은 이름의 폴더가 존재한다면
    print("이미 존재하는 폴더입니다.")
    os.rmdir(folder) # 폴더 삭제
    print(folder, "폴더를 삭제했습니다.")
else: # 폴더가 존재하지 않으면
    os.makedirs(folder) # 폴더 생성
    print(folder, "폴더를 생성했습니다.")

print(os.listdir())
```

```python
# 11.8.3 시간 관련 모듈
# time 모듈
# localtime(): 현재 시간 정보 반환

import time
print(time.localtime())
```

```python
# strftime(): 사용자가 원하는 문자열 형태로 시간 정보 출력 가능
# %Y-연 / %m-월 / %d-일 / %H-시 / %M-분 / %S-초

# ex. 2026-11-06 15:00:00처럼 출력하려면...
import time
print(time.strftime("%Y-%m-%d %H:%M:%S")) # 연-월-일 시:분:초
```

```python
# datetime 모듈
# 오늘 날짜 출력

import datetime
print("오늘 날짜는", datetime.date.today())
```

```python
# datetime 모듈 - timedelta() 함수
# 두 날짜 및 시간 사이의 차이를 계산
# 또는 일정 시간이 경과한 후의 날짜를 구할 수 있음

import datetime

today = datetime.date.today() # 오늘 날짜 저장
td = datetime.timedelta(days=100) # 100일째 날짜 저장
print("우리가 만난 지 100일은", today + td) # 오늘부터 100일 후 날짜
```
