```python
# 문제. 사이트별로 비밀번호를 생성하는 프로그램을 작성하세요.
# http://naver.com
# http://daum.net
# http://google.com
# http://youtube.com

# 조건1. http:// 부분은 제외한다.
# 조건2. 처음 만나는 점(.)이후 부분도 제외한다.
# 조건3. 남은 글자 중 처음 세 자리 + 글자 개수 + 글자 내 'e'의 개수 + '!'로 구성한다.

# 조건4. 프로그램을 실행했을 때 실행 결과는 다음 형태로 나와야 한다.
# http://naver.com의 비밀번호는 nav51!입니다.
```

```python
# 1. http://부분 제외  naver.com
# 2. 처음 만나는 점 이후 부분 제외 一 naver
# 3. 남은 글자 중 처음 세 자리(nav) + 글자 개수(5)十 글자 내 ‘e’의 개수⑴ + (!)
```

```python
# 조건1. http:// 부분은 제외한다.
site = "http://naver.com" 
site2 = site[7:]

# 조건2. 처음 만나는 점(.)이후 부분도 제외한다.
find = site2.find(".")
print(find)

name = site2[:find]
print(name)

# # 조건3. 남은 글자 중 처음 세 자리 + 글자 개수 + 글자 내 'e'의 개수 + '!'로 구성한다.
p1 = name[:3]
p2 = str(len(name))
p3 = str(name.count("e"))
password = p1 + p2 + p3 + '!'
print(password)
```

```python
# 조건4. 프로그램을 실행했을 때 실행 결과는 다음 형태로 나와야 한다.
# http://naver.com의 비밀번호는 nav51!입니다.

print(f"{site}의 비밀번호는 {password}입니다.")
```

```python
# 해설
url = "http://naver.com"

my_str = url.replace("http://", "") # 조건1, 앞의 http:// 부분을 없앰
my_str = my_str[:my_str.index(".")] # 조건2, . 나오는 부분까지
password = my_str[:3] + str(len(my_str)) + str(my_str.count("e")) + "!" # 조건3, 조합
print("{0}의 비밀번호는 {1}입니다.".format(url, password))
```

```python
# 문제. 영어 문장이 주어졌을 때 첫 번째 글자는 대문자로,
# 나머지 글자는 모두 소문자로 변환하는 프로그램을 작성하세요.

# 문장1. the early bird catches the worms.
sentence1 = "the early bird catches the worms."
print(sentence1[0].upper() + sentence1[1:].lower())

# 문장2. Actions Speak Louder Than Words.
sentence2 = "Actions Speak Louder Than Words."
print(sentence2[0].upper() + sentence2[1:].lower())

# 문장3. PRACTICE MAKES PERFECT.
sentence3 = "PRACTICE MAKES PERFECT."
print(sentence3[0].upper() + sentence3[1:].lower())
```

```python
# 파이썬에서 제공하는 문자열 함수: .capitalize()
# 첫 글자는 대문자로, 나머지는 소문자로 변경하는 함수

print(sentence1.capitalize())
print(sentence2.capitalize())
print(sentence3.capitalize())
```
