```python
# 8.4.1 파일 열고 닫기: open(), close()

# open() 형식
# open("파일명", "모드", encoding="인코딩 형식")

# 모드 종류
# r = read 읽기 전용
# w = write 쓰기 / 파일에 내용을 쓰기 위한 모드, 같은 이름의 파일이 있으면 해당 파일을 덮어쓰고 기존 내용 삭제
# a = append 이어 쓰기 / 같은 이름의 파일이 있으면 기존 내용 끝에 이어씀

# 인코딩
# 파일 내용에 담긴 문자 표시와 관련한 내용
# 값은 utf8로 설정하면 한글을 포함한 파일을 사용할 때도 문제 X
```

```python
# 학교 성적 정보를 담은 텍스트 파일 예제

score_file = open("score.txt", "w", encoding="utf8") # score.txt 파일을 쓰기 모드로 열기
print("수학 : 0", file=score_file) # score.txt 파일에 내용 쓰기
print("영어 : 50", file=score_file) # score.txt 파일에 내용 쓰기
score_file.close() # score.txt 파일 닫기
# 내용 저장 및 충돌 방지 위해 close() 함수 꼭 사용
```

```python
# 8.4.2 파일 쓰기: write()

score_file = open("score.txt", "a", encoding="utf8") # score.txt 파일을 이어쓰기 모드로 열기
score_file.write("과학 : 80\n") # write 함수는 줄바꿈이 없으므로 \n 추가
score_file.write("코딩 : 100\n")
score_file.write("국어 : 40\n")
score_file.close()
```

```python
# 8.4.3 파일 읽기: read(), readline(), readlines()

# read() 한 번에 내용 읽어올 때
score_file = open("score.txt", "r", encoding="utf8") # score.txt 파일을 읽기 모드로 열기
print(score_file.read()) # 파일 전체 읽어오기
score_file.close()
```

```python
# readline() 한 줄씩 읽을 때
score_file = open("score.txt", "r", encoding="utf8")
print(score_file.readline(), end="") # 한 줄씩 읽어 오고 커서는 다음 줄로 이동
print(score_file.readline(), end="") # end 값을 비움으로 설정해 줄 바꿈 중복 수행 방지
print(score_file.readline(), end="")
print(score_file.readline(), end="")
print(score_file.readline(), end="")
score_file.close()
```

```python
# 줄 수를 아는 경우는 그만큼 코드를 작성하면 되지만,
# 파일 내 몇 줄인지 알 수 없을 땐 while() 함수 사용
# 파일에 내용이 남아있는 동안 반복해서 읽어오고 끝나면 탈출

score_file = open("score.txt", "r", encoding="utf8")

while True:
    line = score_file.readline()
    if not line: # 더 이상 읽어 올 내용이 없을 때
        break
    print(line, end="") # 읽어온 내용 출력

score_file.close()
```

```python
# 파일 내용을 한꺼번에 불러와서 리스트에 저장하고 리스트를 반복하면서 내용 출력하는 경우
# readlines() 함수 사용
# 보통 for 문과 함께 사용

score_file = open("score.txt", "r", encoding="utf8")

lines = score_file.readlines() # 파일에서 모든 줄을 읽어와 리스트 형태로 저장
for line in lines: # lines에 내용이 있을 때까지
    print(line, end="") # 읽어온 내용 출력

score_file.close()
```

```python
# read() : 파일 내용 한꺼번에 읽어오기
# readline() : 파일 내용 한 줄씩 읽어오기 + 주로 while과 사용
# readlines() : 파일 내용을 줄 단위로 나뉜 리스트 형태로 한꺼번에 읽어오기 + 주로 for와 사용
```
