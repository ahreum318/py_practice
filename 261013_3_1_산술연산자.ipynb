```python
# 3.1.1 산술 연산자

# + : 더하기
# - : 빼기
# * : 곱하기
# / : 나누기

print(1 + 1)
print(3 - 2)
print(5 * 2)
print(6 / 3)
```

```python
# ** : 거듭제곱
# % : 나눈 나머지
# // : 정수 나눗셈 (소수점 내림)

print(2 ** 3)
print(10 % 3)
print(10 // 3)
print(-7 // 2) # 버림이 아닌 내림 처리라 출력 결과 : -4
```

```python
# 3.1.2 비교 연산자
# > : 왼쪽이 큼
# >= : 왼쪽이 오른쪽보다 크거나 같음
# < : 오른쪽이 큼
# <= : 왼쪽이 오른쪽보다 작거나 같음

print(10 > 3)
print(4 >= 7)
print(10 < 3)
print(5 <= 5)
```

```python
# == : 왼쪽과 오른쪽이 같음 # =은 파이썬에서는 대입 연산자
# != : 왼쪽과 오른쪽이 다름

print(3 == 3)
print(4 == 2)
print(3 + 4 == 7)
print(1 != 3)
```

```python
# 3.1.3 논리연산자
# and = 왼쪽과 오른쪽이 모두 참이면 참
# or = 왼쪽과 오른쪽 중 하나라도 참이면 참
# not = 값이 참이면 거짓, 거짓이면 참

print((3 > 0) and (3 > 5))
print((3 > 0) or (3 > 5))
print(not(1 != 3))
```
