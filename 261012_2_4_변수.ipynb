```python
# 변수 = 어떤 값을 저장하는 공간
# 등호 (=) 로 값 대입
# = 대입연산자

# 변수명 = 값

name = "연탄이"
animal = "개"
age = 4
hobby = "산책"
is_male = True

# ㄴ위와 같은 과정을 변수를 정의한다고 표현
```

```python
# 변수명 규칙
# 변수명에는 소문자(a-z), 대문자(A-Z), 숫자(0-9), 언더바(_)를 사용
# 변수명은 대소문자를 구분: Apple, apple, APPLE 전부 다른 변수
# 변수명은 숫자로 시작할 수 없고 소문자/대문자/언더바로 시작할 수 있음
# 키워드(파이썬에서 사용이 예약된 문자열)는 변수명 사용 X
```

```python
# 형식: "문자열" + 변수 + "문자열"

# SAMPLE
print("반려동물을 소개해 주세요.")
print("우리 집 반려동물은 개인데, 이름이 연탄이예요.")
print("연탄이는 4살이고, 산책을 아주 좋아해요.")
```

```python
print("반려동물을 소개해 주세요.")
print("우리 집 반려동물은 " + animal + "인데, 이름이 " + name + "예요.")
```

```python
# print(name + "는 " + age + "살이고, " + hobby + "을 아주 좋아해요.")
# ㄴ이 경우 문자열, 숫자가 결합되어서 오류 발생
# ㄴ + 로 연결할 때는 자료형이 같아야 함
# can only concatenate str (not "int") to str

print(name + "는 " + str(age) + "살이고, " + hobby + "을 아주 좋아해요.")
# str()은 값의 자료형을 문자열로 바꾸는 명령어
# 형변환 (2.4.3)
```

```python
# 쉼표로 연결도 가능
# 단 +로 연결할 때와 2가지 다름

# 1. str() 형변환 없이도 연결 가능
print(name, "는", age, "살이고", hobby, "을 아주 좋아해요.")

# 2. 쉼표 부분 빈칸 삽입됨
# 해피 는 4 살이고 낮잠 을 아주 좋아해요.
```

```python
# 형변환1. 정수형 int()
print(int("3"))
print(int(3.5)) # 실수를 정수로 변환할 땐 소수점 버림
```

```python
print(int("3") + "입니다.")
# 정수형 + 문자열 조합이라 에러 발생
```

```python
print(int("삼"))
# 문자열을 int()로 형변환 할 때는 오로지 숫자로 된 문자열만 사용 가능
```

```python
# 형변환2. 실수형 float()
print(float("3.5"))
print(float(3)) # 소수점 이하 첫째 자리에 0을 붙여 실수 3.0으로 변환
```

```python
print(float("오"))
# int()와 float()는 모두 숫자로 된 문자열에만 사용 가능
```

```python
# 형변환3. str()
print(str(3) + "입니다.")
print(str(3.5) + "입니다.")
```

```python
# type()
# 데이터의 자료형 확인할 때 type() 사용.
# 소괄호 안에 확인하고 싶은 데이터를 넣고 출력

print(type(3))
print(type("3"))
print(type(3.5))
print(type(str(3)))
```
