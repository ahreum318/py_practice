```python
number = 2 + 3 * 4
print(number)
```

```python
# number에 2를 더하려면
number = number + 2
print(number)
```

```python
# 복합 대입 연산자

num = 2
num += 2 # num = num + 2
print (num)
```

```python
num -= 2 # num = num - 2
num *= 2# num = num * 2
num /= 2 # num = num / 2
num **= 2# num = num ** 2
num %= 2 # num = num % 2
num //= 2# num = num // 2
```
