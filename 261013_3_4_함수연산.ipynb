```python
# 3.4.3 random 모듈
from random import *

print(random())
print(random())
print(random())

# random() = 0이상 1미만 사이의 랜덤 난수 출력
```

```python
print(random() * 10) # 0.0 이상 10.0 미만에서 난수 생성
print(int(random() * 10)) # 0이상 10미만 정수에서 난수 생성 (random()결과를 int()로 감싸서 정수 변환)
print(int(random() * 10) + 1) # 1이상 11미만 정수에서 난수 생성
```

```python
# 로또 번호 뽑기 작성
print(int(random() * 45) + 1)
```

```python
# random 모듈의 함수
randrange(시작 숫자, 끝 숫자)  # 주어진 범위 안에서 정수인 난수 생성(끝 숫자 미포함)
randint(시작 숫자, 끝 숫자)   # 주어진 범위 안에서 정수인 난수 생성(끝 숫자 포함)
```

```python
print(randrange(1, 46))
print(randint(1, 45))
# 둘의 출력 범위는 같음
```

```python
print(randint(1, 45))
print(randint(1, 45))
print(randint(1, 45))
print(randint(1, 45))
print(randint(1, 45))
print(randint(1, 45))

# 로또 번호 뽑는 법?
# 이렇게 하면 안된다! 각 문장은 독립된 문장이어서 중복 발생할 수 있음
# 헐대박마침 중복나옴

```
