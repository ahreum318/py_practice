```python
# 문제. 연산자를 이용해 온도 단위를 변환하는 프로그램을 만들어 보세요.

# 조건1. 섭씨 온도를 저장하기 위한 변수를 만든다.
# 조건2. 다음 공식을 이용해 섭씨 온도를 화씨 온도로 변환하고 새로운 변수에 저장한다.
# ㄴ화씨 온도 = (섭씨 온도 * 9 / 5) + 32

celsius = 10
fahrenheit = (celsius * 9 / 5) + 32

print("섭씨 온도 : " + str(celsius))
print("화씨 온도 : " + str(fahrenheit))
```
