```python
lower() #문자열을 소문자로 변환
upper() #문자열을 대문자로 변환
islower() #문자열이 소문자인지 확인
isupper() #문자열이 대문자인지 확인
replace() #문자열 바꾸기
index() #찾는 문자열의 인덱스(없으면 오류 발생)
find() #찾는 문자열의 인덱스(없으면 -1 반환)
count() #문자열이 나온 횟수
```

```python
#문자열 처리 함수의 형식
문자열(또는 변수).함수()
```

```python
python = "Python is Amazing"

print(python.lower())
print(python.upper())
print(python[0].isupper()) #인덱스 0에 있는 값이 대문자인지 확인
print(python[1:3].islower()) #인덱스 1부터 2에 있는 값이 소문자인지 확인
print(python.replace("Python", "Java"))
```

```python
# 문자열에서 어떤 문자를 찾는 함수
# find(), index()

find(찾는 문자, 시작 인덱스, 종료 인덱스)
index(찾는 문자, 시작 인덱스, 종료 인덱스)
# 시작/종료 인덱스는 생략 가능함
# 둘 다 생략 시 문자열 전체에서 찾음
```

```python
python = "Python is Amazing"

find = python.find("n") # 처음 발견한 n의 인덱스
print(find)
```

```python
find = python.find("n", find + 1) # 인덱스 6 이후부터 찾아 처음 발견한 n의 인덱스
print(find)
```

```python
find = python.find("Java")
print(find)
```

```python
python = "Python is Amazing"

index = python.index("n") # 처음 발견한 n의 인덱스
print(index)
```

```python
index = python.index("n", index + 1) # 인덱스 6 이후부터 찾아 처음 발견한 n의 인덱스
print(index)
```

```python
find = python.index("Java")
print(index)
```

```python
python = "Python is Amazing"

print(python.count("n"))
print(python.count("v"))
```

```python
# len()
# 문자열의 길이 정보 알려줌(공백 포함)

python = "Python is Amazing"
print(len(python))
```
