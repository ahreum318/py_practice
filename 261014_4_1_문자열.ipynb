```python
# 문자열 = 문자들의 집합
# 파이썬에서는 한글이나 알파벳 등으로 작성한 글자를 '' 또는 ""로 감싸서 문자열임을 나타냄

sentence1 = '나는 소년입니다.'   # 얘가 문자열
print(sentence1)
```

```python
sentence2 = "파이썬은 쉬워요"
print(sentence1, type(sentence2))
print(sentence2, type(sentence2))
```

```python
# 문자열을 여러 줄에 걸쳐 작성해야 하는 경우
# """
# """ 이렇게 감싸기

sentence3 = """
나는 소년이고,
파이썬은 쉬워요.
"""

print(sentence3, type(sentence3))
```
