```python
# 슬라이싱: 원하는 만큼 데이터를 자를 수 있음
# 인덱스: 여러 문자로 구성된 문자열의 n번째 문자(또는 데이터)가 있을 때 n번째 = 인덱스
# ㄴ즉, 인덱스는 데이터의 순서 또는 위치를 나타냄

# index
# 저장한 변수명에 []를 붙이고 그 안에 숫자를 넣어 표시함
# 형식: 변수명[인덱스]
```

```python
jumin = "990229-1234567"
print("성별 식별번호 : " + jumin[7])

# 성별을 나타내는 1이란 값은 8번째 자리에 있음, 그러나 인덱스는 7
# ★ 코드에서 위치를 나타내는 인덱스는 1이 아닌 0부터 시작함
```

```python
# 슬라이싱
# 형식: 변수명[시작 인덱스:종료 인덱스]
# [3:7] = 인덱스 3부터 7 직전(★ 6까지)

jumin = "990229-1234567"
print("연 : " + jumin[0:2]) # (0, 1)
print("월 : " + jumin[2:4]) # (2, 3)
print("일 : " + jumin[4:6]) # (4, 5)
```

```python
# 시작 인덱스나 종료 인덱스를 비워두면 비워둔 위치에 따라 슬라이싱 범위가 달라짐

# 변수명[:종료 인덱스] : 처음부터 종료 인덱스 직전까지 슬라이싱
# 변수명[시작 인덱스:] : 시작 인덱스부터 끝까지 슬라이싱
# 변수명[:] : 처음부터 끝까지 슬라이싱

print("생년월일 : " + jumin[:6]) # jumin[0:6]
print("주민등록번호 뒷자리 : " + jumin[7:]) # jumin[7:14]
```

```python
# 뒤에서부터 슬라이싱도 가능
# 이때는 음수 인덱스 사용
# ★★ 단, 양수 인덱스는 0부터 시작하지만 음수 인덱스일 때부터는 -1부터 시작함
# 7 : -1, 6 : -2, 5 : -3, 4: -4 ... 9 : -14

print("주민등록번호 뒷자리(뒤에서부터) : " + jumin[-7:]) # 뒤에서 7번째 위치부터 끝까지
```
