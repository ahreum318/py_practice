```python
# 5.2.1 딕셔너리 생성하기
# 딕셔너리 구조: key와 value 한 쌍
# 형식: 딕셔너리명 = {key1: value1, key2: value2, ...}

# key는 중복을 허용하지 않음
# key는 변하지 않는 값을 사용해야 함
```

```python
cabinet = {3: "푸", 100: "피글렛"}

# 빈 딕셔너리 생성 시
# empty_dict = {}

# 딕셔너리에서는 인덱스 역할을 key가 함
print(cabinet[3])
print(cabinet[100])
print(cabinet.get(3)) # get 함수 사용 가능
```

```python
# 단, get함수는 정의되지 않은 key 전달 시 None 반환
# 대괄호 사용 경우에는 오류 발생

print(cabinet.get(5)) # None 반환
print("hi")
print(cabinet[5]) # key = 5 정의되지 않아 오류 발생
print("hi")
```

```python
# get으로 기본값 설정도 가능
# key = 5를 정의하기 전까지는 정의한 value로 나옴

print(cabinet.get(5, "사용 가능"))

# get 함수의 기본 형식: get(key, default=None)
# key에 해당하는 value가 없으면 None 대신 "사용 가능"을 반환해라
```

```python
# 자료구조에 해당 key가 있는지 확인하기 = in

print(3 in cabinet)
print(5 in cabinet)

# 문자열에 해당 글자가 포함됐는지 확인할 때도 사용 가능
print("곰" in "곰돌이")
print("돌이" in "곰돌이")
print("푸" in "곰돌이")
```

```python
# key에는 문자열도 포함 가능
cabinet = {"A-3": "푸", "B-100": "피글렛"}
print(cabinet["A-3"])
print(cabinet["B-100"])
```

```python
# 5.2.2 값 변경/추가/삭제
cabinet = {"A-3": "푸", "B-100": "피글렛"}
print(cabinet)
cabinet["A-3"] = "티거" # key에 해당하는 값이 있을 때 > 값 변경
cabinet["C-20"] = "이요르" # key에 해당하는 값이 없을 때 > 값 추가
print(cabinet)

del cabinet["A-3"] # key 'A-3'에 해당하는 값 삭제
print(cabinet)
```

```python
# 5.2.3 값 확인하기

# 현재 딕셔너리 안에 있는 key값 확인하고 싶을 때, keys() 함수 사용
print(cabinet.keys())

# value값 확인 시 values() 함수 사용
print(cabinet.values())

# key-value값 같이 확인하고 싶을 땐 items() 함수 사용
print(cabinet.items())
```

```python
# 모든 값 삭제 시 clear() 함수 사용
# 리스트와 동일

cabinet.clear()
print(cabinet)
```
