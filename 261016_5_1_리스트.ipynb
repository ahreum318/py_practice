```python
# 5.1.1 리스트 생성하기
# 관련 변수가 여러 개일 때 리스트 사용 시 변수 하나로 관리 가능

# 형식 : 리스트명 = [값1, 값2, ...]

subway = [10, 20, 30]
print(subway)
```

```python
# 5.1.2 값 추가/삽입/삭제

subway = ["푸", "피글렛", "티거"]
print(subway)
```

```python
# 리스트에도 인덱스 존재
print(subway.index("피글렛"))
```

```python
# 값 추가 시 append() 함수 사용
# 자동으로 맨 끝에 값 추가
subway.append("이요르")
print(subway)

# 중간에 값을 추가하고 싶다면 insert() 함수 사용
# 형식: insert(인덱스, 삽입할 값)
subway.insert(1, "루")
print(subway)
```

```python
# 값 삭제 시 pop() 함수 사용
# 리스트 끝에서부터 값을 꺼내고, 해당 값 삭제

print(subway.pop())
print(subway)
```

```python
print(subway.pop())
print(subway)
```

```python
print(subway.pop())
print(subway)
```

```python
# 리스트에서 값이 더 이상 필요없을 때, 새로운 값을 저장하고 싶을 때
# clear() 함수로 리스트의 모든 값 삭제 가능

subway.clear()
print(subway)
```

```python
# 5.1.3 중복 값 확인하기

subway = ["푸", "피글렛", "티거"]
subway.append("푸")
print(subway)
print(subway.count("푸"))
```

```python
# 5.1.4 리스트 정렬하기
# sort() 함수 사용 시 정렬 가능
# 디폴트: 오름차순 sort(reverse=False) = sort()
# 내림차순 sort(reverse=True)

num = [5, 2, 4, 3, 1]
# print(num.sort())  sort()는 반환값을 따로 주지 않는 함수라 이렇게 쓰면 None 뱉음
num.sort(reverse=True)
print(num)

# reverse() 함수 사용 시 순서 뒤집기 가능
num.reverse()
print(num)
```

```python
# sorted()로 리스트 정렬하기
# sort()는 원본값 직접 변경하지만, sorted()는 원본 리스트를 정렬해서 새로운 값을 생성

my_list = [1, 3, 2]
my_list.sort() # 리스트 정렬
print(my_list) # my_list의 데이터 직접 변경

your_list = [1, 3, 2]
new_list = sorted(your_list)
print(your_list) # 원본 리스트 데이터 유지
print(new_list) # 새로운 리스트 생성
```

```python
# 5.1.5 리스트 확장하기
# 리스트에 같은 자료형의 값만 넣을 필요 x
# 정수형, 실수형, 문자열, 불형, 리스트까지 집어넣기 가능

mix_list = ["푸", 20, True, [5, 2, 4, 3, 1]]
print(mix_list)
```

```python
# extend() 함수로 서로 다른 리스트를 합칠 수 있음
# 형식: 리스트1.extend(리스트2)

mix = ["푸", 20, True]
num = [5, 2, 4, 3, 1]
num.extend(mix)
print(mix)
print(num)
```

```python
# 정렬은 비교가 불가하면 안됨
# ex. 정수형과 실수형 < 가능
# ex. 정수형과 문자열 < 불가

num.sort()
print(num)
```
