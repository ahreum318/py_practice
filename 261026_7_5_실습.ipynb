```python
# 문제. 표준 체중을 구하는 프로그램을 작성하세요.

# 표준 체중: 키에 따른 평균 체중
# 성별에 따른 표준 체중 공식
# 남자: 키(m) x 키(m) x 22
# 여자: 키(m) x 키(m) x 21

# 조건1. 표준 체중은 별도 함수로 계산한다. 단, 키는 미터(m) 단위로 받는다.
# 함수명: std_weight
# 전달값: 키(height), 성별(gender)

# 조건2. 실행결과에서 표준 체중은 소수점 이하 둘째 자리까지 표시한다.
# 실행결과: 키 175cm 남자의 표준 체중은 67.38kg입니다.
```

```python
def std_weight(height, gender):
    if gender == "남자":
        weight = height * height * 22
    elif gender == "여자":
        weight = height * height * 21
    else:
        print("오류 발생! 성별을 재입력 해주세요.")
        return

    print(f"키 {int(height * 100)}cm {gender}의 표준 체중은 {weight:.2f}kg 입니다.")

std_weight(1.75, "남자")
std_weight(1.60, "여자")
std_weight(1.70, "외계인")
```

```python
# 해설
def std_weight(height, gender):
    if gender == "남자":
        return height * height * 22
    else:
        return height * height * 21

height = 175
gender = "남자"
weight = round(std_weight(height / 100, gender), 2)
print("키 {0}cm {1}의 표준 체중은 {2}kg입니다.".format(height, gender, weight))
```

```python
# 문제. 미세먼지 수치를 입력받아 대기질 상태를 출력하는 함수를 만드시오.

# 조건1. get_air_quality라는 이름으로 함수를 만든다.
# 조건2. 이 함수는 전달값으로 미세먼지 수치를 입력받는다.
# 조건3. 이 함수는 대기질 상태를 반환한다.

# 조건4. 미세먼지 수치에 따른 대기질 상태는 다음과 같다.
# 좋음: 0~30
# 보통: 31~80
# 나쁨: 81~150
# 매우 나쁨: 151 이상

# 조건 5. 함수에 전달되는 전달값은 항상 0 이상의 값이라고 가정한다.

# 예시
# print(get_air_quality(15)) # 좋음
# print(get_air_quality(85)) # 나쁨
```

```python
def get_air_quality(dust):
    if 0 <= dust <= 30:
        return "좋음"
    elif dust <= 80:
        return "보통"
    elif dust <=150:
        return "나쁨"
    else:
        return "매우 나쁨"


print(get_air_quality(47))

# 해설도 동일!
```
