```python
# 문자 입력된 경우
print("나누기 전용 계산기입니다.")
num1 = int(input("첫 번째 숫자를 입력하세요 : "))
num2 = int(input("두 번째 숫자를 입력하세요 : "))
print("{0} / {1} = {2}".format(num1, num2, int(num1 / num2)))
```

```python
# 위 경우 문자열의 예외 필요

# 예외 처리 형식
# try:
    # 실행 명령1
    # 실행 명령2
    # ...
# except 오류 종류:
    # 예외 처리 명령1
    # 예외 처리 명령2
    # ...

try:
    print("나누기 전용 계산기입니다.")
    num1 = int(input("첫 번째 숫자를 입력하세요 : "))
    num2 = int(input("두 번째 숫자를 입력하세요 : "))
    print("{0} / {1} = {2}".format(num1, num2, int(num1 / num2)))
except ValueError:
    print("오류 발생! 잘못된 값을 입력했습니다.")
```

```python
# 10.1.3 오류 메시지를 예외 처리로 출력하기: as
# 0으로 나눌 경우...

try:
    print("나누기 전용 계산기입니다.")
    num1 = int(input("첫 번째 숫자를 입력하세요 : "))
    num2 = int(input("두 번째 숫자를 입력하세요 : "))
    print("{0} / {1} = {2}".format(num1, num2, int(num1 / num2)))
except ValueError:
    print("오류 발생! 잘못된 값을 입력했습니다.")
```

```python
try:
    print("나누기 전용 계산기입니다.")
    num1 = int(input("첫 번째 숫자를 입력하세요 : "))
    num2 = int(input("두 번째 숫자를 입력하세요 : "))
    print("{0} / {1} = {2}".format(num1, num2, int(num1 / num2)))
except ValueError:
    print("오류 발생! 잘못된 값을 입력했습니다.")
except ZeroDivisionError as err: # 오류 메시지를 err라는 변수 안에 담아 출력
    print(err)
```

```python
try:
    print("나누기 전용 계산기입니다.")
    nums = []
    nums.append(int(input("첫 번째 숫자를 입력하세요 : ")))
    nums.append(int(input("두 번째 숫자를 입력하세요 : ")))
    # nums.append(int(nums[0] / nums[1])) # 연산 결과를 리스트에 추가
    print("{0} / {1} = {2}".format(nums[0], nums[1], nums[2]))

except ValueError:
    print("오류 발생! 잘못된 값을 입력했습니다.")

except ZeroDivisionError as err: # 오류 메시지를 err라는 변수 안에 담아 출력
    print(err)

except Exception as err: # 정의하지 않은 모든 오류를 예외처리
    print("알 수 없는 오류가 발생했습니다.")
    print(err)
```

```python
# 예외 처리 클래스

ValueError = 값 에러
ZeroDivisionError = 어떤 숫자를 0으로 나눌 수 없음
IndexError = 인덱스 에러
NameError = 변수명 없음
SyntaxError = 문법 오류
FileNotFoundError = 접근하려는 위치에 파일 없음
# 의 부모클래스
Exception
```
