```python
# 내장 함수: 별도로 import하지 않고도 사용할 수 있는 함수
# input(), print()도 내장 함수 중 하나

language = input("어떤 언어를 좋아하세요? ")
print("{0}은 아주 좋은 언어입니다!".format(language))
```

```python
# dir()
# 어떤 객체를 전달값으로 넘기면 이 객체가 어떤 변수와 함수를 가지고 있는지 알려줌
# 전달값으로 아무것도 넣지 않으면 현재 소스 코드 안에서 사용할 수 있는 모듈 또는 객체 출력
print(dir())
```
