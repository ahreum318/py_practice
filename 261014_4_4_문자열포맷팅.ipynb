```python
# 원하는 위치에 특정한 값(또는 변수)을 넣어서 하나의 문자열로 표현하는 방법 : 문자열 포매팅
# 여러가지 방법이 있음
```

```python
# 4.4.1 서식 시정자(format specifier)
# 자료형을 표현하는 방법으로 %뒤에 자료형을 나타내는 문자가 붙음

# %d : 정수
# %f : 실수
# %c : 문자
# %s : 문자열

# 형식 : print("문자열 서식지정자 문자열" % 값)
```

```python
print("나는 %d살입니다." % 20)
print("나는 %s을 좋아합니다." % "파이썬")
print("Apple은 %c로 시작해요." % "A")
print("나는 %s살입니다." % 20) # %s로도 정수 표현 가능
print("나는 %s색과 %s색을 좋아해요." % ("파란", "빨간")) # 값이 여러 개일 때
```

```python
# 4.4.2 format() 함수
# 형식: print("문자열{인덱스}문자열{인덱스}...".format(값1, 값2, ...))

print("나는 {}살입니다.".format(20))
print("나는 {}색과 {}색을 좋아해요.".format("파란", "빨간"))
print("나는 {0}색과 {1}색을 좋아해요.".format("파란", "빨간"))
print("나는 {1}색과 {0}색을 좋아해요.".format("파란", "빨간"))
```

```python
# 중괄호에 이름을 넣고 format() 함수에 이름과 값을 정의해도 됨

print("나는 {age}살이며, {color}색을 좋아해요".format(age=20, color="빨간"))
print("나는 {age}살이며, {color}색을 좋아해요".format(color="빨간", age=20))
```

```python
# ★ 4.4.3 f-문자열 사용하기
# 문자열 앞에 f를 추가하면 문자열이 나오기 전에 정의한 변수의 값을 사용 가능

age = 20
color = "빨간"
print(f"나는 {age}살이며, {color}색을 좋아해요.")
```

```python
apple = "사과"
banana = "바나나"

print(f"빨가면 {apple} 맛있으면 {banana}")
```
