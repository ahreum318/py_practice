```python
# 6.1.1 조건이 하나일 때: if 문

# 형식
# if 조건:
#     실행할 명령

weather = "비"

if weather == "비":
    print("우산을 챙기세요.") #들여쓰기 공백 4칸. 공백 개수가 달라지면 코드 작동 X
```

```python
# 6.1.2 조건이 여러 개일 때: elif 문

# 형식
# if 조건1:
    # 실행할 명령1
# elif 조건2:
    # 실행할 명령2
# elif 조건3:
    # 실행할 명령3
# ...

weather = "미세먼지"
if weather == "비":
    print("우산을 챙기세요.")
elif weather == "미세먼지":
    print("마스크를 챙기세요.")
```

```python
# 6.1.3 모든 조건에 맞지 않을 때: else 문

# 형식
# if 조건1:
    # 실행할 명령1
# elif 조건2:
    # 실행할 명령2
# elif 조건3:
    # 실행할 명령3
# ...
# else:
    # 실행할 명령n

weather = "맑음"
if weather == "비":
    print("우산을 챙기세요.")
elif weather == "미세먼지":
    print("마스크를 챙기세요.")
else:
    print("준비물이 필요 없어요.")
```

```python
# 6.1.4 input()으로 값 입력받기

# input(): 어떤 값을 입력받는 용도
# 값은 항상 문자열 상태로 변수에 저장됨
# 숫자를 입력해도 문자열 "3"으로 인식

weather = input("오늘 날씨는 어때요?")
print(weather)
```

```python
weather = input("오늘 날씨는 어때요?")

if weather == "비" or weather == "눈": # or 조건
    print("우산을 챙기세요.")
elif weather == "미세먼지":
    print("마스크를 챙기세요.")
else:
    print("준비물이 필요 없어요.")
```

```python
temp = int(input("오늘 기온은 어때요?")) #문자열로 저장된 입력값을 정수형으로 변환

if 30 <= temp:
    print("너무 더워요. 외출을 자제하세요.")
elif 10 <= temp and temp < 30:
    print("활동하기 좋은 날씨예요.")
elif 0 <= temp and temp < 10:
    print("외투를 챙기세요.")
else:
    print("너무 추워요. 나가지 마세요.")
```
