```python
# pickle 모듈 : 프로그램에서 사용하는 데이터를 파일 형태로 저장하거나 불러올 수 있게 하는 모듈
# pickle 모듈로 데이터를 파일에 저장할 때 dump()라는 함수 사용

# dump() 형식
# dump(저장할 데이터, 저장할 파일명)
```

```python
import pickle # pickle 모듈 가져오기

# pickle 모듈로 저장하는 파일은 바이너리(binary 형태)
# txt: 사람이 읽을 수 있는 글자로 이루어진 파일. 글꼴, 글자 크기, 색상 등 서식 정보 없이 단순한 글자만 저장 가능
# binary: 컴퓨터가 인식할 수 있는 이진수(0, 1)로 이루어진 파일로, jpg/png 같은 이미지 파일, mp3 등의 음악 파일, exe 등의 실행 파일 해당

# 파일 열기 모드
# t : 텍스트 모드로 열기. 읽기 모드일 때는 rt, 쓰기 모드일 때는 wt. 텍스트 모드는 기본값이므로 t는 생략 가능
# b : 바이너리 모드로 열기. 읽기 모드일 때는 rb, 쓰기 모드일 때는 wb.
# 바이너리 모드는 한글이 포함됐더라도 별도의 encoding 필요 없음 (생략 가능)

profile_file = open("profile.pickle", "wb")
profile = {"이름": "스누피", "나이": 30, "취미": ["축구", "골프", "코딩"]}
print(profile)

pickle.dump(profile, profile_file) # profile 데이터를 파일에 저장
profile_file.close() # 파일 닫기
```

```python
# 만든 파일을 불러올 때는 load() 함수 사용
# 형식: load(불러올 파일명)

profile_file = open("profile.pickle", "rb")
profile = pickle.load(profile_file) # 파일에 있는 정보를 불러와서 profile에 저장
print(profile)
profile_file.close()
```
