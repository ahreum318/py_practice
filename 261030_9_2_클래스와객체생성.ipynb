```python
# 클래스
# 흔히 붕어빵 틀에 비유

# 메서드: 클래스 안에 정의하는 함수
# 일반 함수와 다르게 첫 번째 전달값 위치에는 self라고 적음

# 형식
# class 클래스명:
    # def 메서드명1(self, 전달값1, 전달값2, ...):
        # 실행할 명령1
        # 실행할 명령2
        # ...
    # def 메서드명2(self, 전달값1, 전달값2, ...):
        # 실행할 명령1
        # 실행할 명령2
        # ...
```

```python
# 메서드 안에 정의한 변수: 인스턴스 변수
# 형식: self.변수명 = 값

class Unit:
    def __init__(self, name, hp, damage):
        self.name = name # 인스턴스 변수 name에 전달값 name 저장
        self.hp = hp # 인스턴스 변수 hp에 전달값 hp 저장
        self.damage = damage # 인스턴스 변수 damage에 전달값 damage 저장
        print("{0} 유닛을 생성했습니다.".format(self.name))
        print("체력 {0}, 공격력 {1}".format(self.hp, self.damage))
```

```python
soldier1 = Unit("보병", 40, 5)
soldier2 = Unit("보병", 40, 5)
tank = Unit("탱크", 150, 35)

# 이렇게 만들어진 유닛을 객체라고 함
# 이렇게 만들어진 객체를 클래스의 인스턴스라고 부름
```

```python
# 9.2.1 생성자: __init__()
# 생성자는 사용자가 따로 호출하지 않아도 객체를 생성할 때 자동으로 호출되는 메서드
# __init__ 이라는 이름으로 메서드를 정의하면 자동으로 생성자가 됨

# 전달값을 3개가 아닌 1개, 또는 2개만 넘기면
soldier3 = Unit("보병, 40") # 오류 발생

# 객체 생성 시에는 self를 제외하고 __init__() 생성자에 정의한 개수만큼 전달값을 넘겨줘야 함
# 클래스에 따로 생성자를 정의하지 않았다면 전달값 없이 클래스명만으로 객체 생성하면 됨
```

```python
# 9.2.2 인스턴스 변수

# 전투기: 공중 유닛, 은폐 불가
stealth1 = Unit("전투기", 80, 5) # 객체 생성, 체력 80, 공격력 5
# 인스턴스 변수 접근
print("유닛 이름 : {0}, 공격력 {1}".format(stealth1.name, stealth1.damage))
# 객체명이 stealth1이므로 인스턴스 변수에는 stealth1.name과 stealth1.damage로 접근해 값을 출력
```

```python
# 은폐 가능
stealth2 = Unit("업그레이드한 전투기", 80, 5)
# 업그레이드한 전투기만을 위한 특별한 인스턴스 변수 정의, 은폐 상태
stealth2.cloaking = True
```

```python
if stealth2.cloaking == True: # 은폐 상태라면
    print("{0}는 현재 은폐 상태입니다.".format(stealth2.name))
```

```python
if stealth1.cloaking == True: # 은폐 상태라면
    print("{0}는 현재 은폐 상태입니다.".format(stealth2.name))
```

```python
# 9.2.3 메서드
# 클래스 내부에 정의한 함수
# 첫번째로 self 작성
# 메서드 안에서 self.로 인스턴스 변수 접근 가능

# 공격할 수 있는 유닛만을 위한 새로운 클래스 정의
class AttackUnit: # 공격 유닛
    def __init__(self, name, hp, damage):
        self.name = name
        self.hp = hp
        self.damage = damage

    def attack(self, location):
        print("{0} : {1} 방향 적군을 공격합니다. [공격력 {2}]" \
              .format(self.name, location, self.damage))

    def damaged(self, damage):
        print("{0} : {1}만큼 피해를 얻었습니다.".format(self.name, self.hp))
        self.hp -= damage
        print ("{0} : 현재 체력은 {1}입니다.".format(self.name, self.hp))
        if self.hp <= 0:
            print("{0} : 파괴됐습니다.".format(self.name))

        
```

```python
#화염방사병 유닛 만들기

flamethrower1 = AttackUnit("화염방사병", 50, 16)
flamethrower1.attack("5시")
```

```python
flamethrower1.damaged(25)
flamethrower1.damaged(25)
```
