```python
# 8.2.1 구분자 넣기: sep

print("파이썬", "자바") # 공백 O
print("파이썬" + "자바") # 공백 X
```

```python
# 쉼표와 + 이외에도 문자열을 구분하는 기호를 사용자가 직접 지정해 출력 가능
# sep 매개변수 사용하면 됨

print("파이썬", "자바")
print("파이썬", "자바", sep=",") # 공백 대신 , 사용

print("파이썬", "자바", "자바스크립트")
print("파이썬", "자바", "자바스크립트", sep=" vs ")
```

```python
# 8.2.2 문장 끝 지정하기: end
# end의 기본값은 \n (줄바꿈)

print("파이썬", "자바", sep=", ", end="? ")
print("무엇이 더 재미있을까요?") # print() 문 끝을 줄 바꿈에 물음표로 변경해 두 문장을 한 줄에 이어서 출력
```

```python
# 8.2.3 출력 위치 지정하기: file

import sys

# file = print() 문의 실행결과를 어디에 출력할지 지정하는 기능
print("파이썬", "자바", file=sys.stdout) # VSCode의 터미널에 결과를 출력하라는 뜻
print("파이썬", "자바", file=sys.stderr) # sys.stderr : 표준 오류, 오류가 발생했을 때 터미널에 오류 메시지를 띄우라는 뜻

# 위 기능은 로그를 남길 때 중요하게 사용
# sys.stdout는 일반적인 내용 / sys.stderr은 오류 발생 시 관련 내용 출력
```

```python
# 8.2.4 공간 확보해 정렬하기: ljust()와 rjust()

scores = {"수학" : 0, "영어" : 50, "코딩" : 100}

for subject, score in scores.items(): # (key, value)
    print(subject, score)
```

```python
scores = {"수학" : 0, "영어" : 50, "코딩" : 100}

for subject, score in scores.items(): # (key, value)
    print(subject.ljust(8), str(score).rjust(4), sep=":")

# 1. 전달값 8에 의해 공간을 총 8칸 확보
# ljust() 함수는 문자열을 왼쪽 정렬
# 과목명을 제외한 나머지 공간은 빈칸으로 출력

# 2. 점수는 숫자이므로 문자열로 바꿔줌
# ★ ljust(), rjust()는 문자열만 정렬 가능
# rjust() 함수는 문자열을 오른쪽 정렬
# 공간 총 4칸 확보, 점수 제외한 칸은 공백 출력

# 3. sep 값에 콜론 지정하여 과목명과 점수 구분
```

```python
# 8.2.5 빈칸 0으로 채우기: zfill()

for num in range(1, 21):
    print("대기번호: " + str(num))
```

```python
for num in range(1, 21):
    print("대기번호: " + str(num).zfill(3)) # 전달하는 숫자만큼 공간 확보 후 문자열 앞의 빈칸을 0으로 채움
```
