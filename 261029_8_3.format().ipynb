```python
print("{0}".format(500))
# format()의 소괄호 안에 넣은 값이 중괄호 위치에 들어가 출력
# 중괄호 부분을 수정하면 다양한 형태로 문자열을 출력 가능
```

```python
# {0}을 {0: >10}으로 바꾸면
# ㄴ 1. 공간 10칸을 확보한 상태에서
# ㄴ 2. 오른쪽 정렬하고
# ㄴ 3. 나머지 공간은 빈칸으로 둠

# 즉, 콜론 뒤에 오는
# 공백 = 빈칸으로 두기 (다른 기호로도 변경 가능)
# > = 오른쪽으로 정렬하기
# 10 = 지정한 만큼(10) 공간 확보하기

print("{0: >10}".format(500))
print("{0: <4}랄라".format("랄라"))
print("{0:엥>8}".format("집에갈래"))
```

```python
# 양수에 + 기호를 붙이는 경우

# 빈칸으로 두기, 오른쪽 정렬, + 기호 붙이기, 공간 10칸 확보
print("{0: >+10}".format(500))
print("{0: >+10}".format(-500))
```

```python
# 금액 3자리마다 쉼표 찍기
print("{0:,}".format(100000000000))
print("{0:+,}".format(100000000000))
print("{0:+,}".format(-100000000000))
```

```python
# 공간 30칸을 확보하고 왼쪽 정렬로 + 기호를 포함한 큰 수를 출력하는데 빈칸은 ^로 채워보자
print("{0:^<+30,}".format(132487124))
```
