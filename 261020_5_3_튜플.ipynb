```python
# 리스트의 '읽기 전용 버전' = 튜플
# 튜플은 처음 정의할 때를 제외하고 값의 변경이나 추가, 삭제가 불가능
# 값의 순서도 바뀔 수 없으나 리스트보다 속도가 빠름

# 형식(소괄호): 튜플명 = (값1, 값2, ...)

menu = ("돈가스", "치즈돈가스")
print(menu[0])
print(menu[1])
```

```python
name = "피글렛"
age = 20
hobby = "코딩"
print(name, age, hobby)

# 튜플로 정의 가능
(name, age, hobby) = ("피글렛", 20, "코딩")
print(name, age, hobby)
```

```python
(departure, arrival) = ("김포", "제주") 
print(departure, ">", arrival) # 김포〉제주 
(departure, arrival) = (arrival, departure) 
print(departure, ">", arrival) # 제주〉김포
```
