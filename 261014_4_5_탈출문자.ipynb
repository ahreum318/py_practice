```python
# 탈출문자
# \와 특정 문자를 함께 사용해 문장 안에서 원하는 동작을 수행하는 방법
```

```python
# 4.5.1 \n: 줄바꿈
print("백문이 불여일견\n백견이 불여일타")
```

```python
# 4.5.2 \" \'

# 저는 "나도코딩"입니다.
# print("저는 "나도코딩"입니다.") # 오류발생 

print("저는 '나도코딩'입니다.")
print('저는 "나도코딩"입니다.')

print("저는 \"나도코딩\"입니다.")
```

```python
# 4.5.3 \\
# print("C：\Users\Nadocoding\Desktop\PythonWorkspace")   # 탈출문자 오류
print("C：\\Users\\Nadocoding\\Desktop\\PythonWorkspace")
print(r"C：\Users\Nadocoding\Desktop\PythonWorkspace")   # r은 문자열 내에 어떤 값이 있든지 무시하고 출력
```

```python
# 4.5.4 \r
# 커서를 맨 앞으로 이동시킴

print("Red Apple\rPine")
# Red Apple 출력 → 커서를 맨 앞으로 이동 → Pine 출력(글자수만큼 insert됨)
print("yellow banana\rhappy")
```

```python
# 4.5.5 \b
# 비파괴 백스페이스, 앞 글자 하나 삭제
print("Redd\bApple")

# 백스페이스와의 차이
print("Banana\b\b\b\b\b\bApple")
```

```python
# 4.5.6 \t
# tab처럼 여러 칸을 띄어쓰는 역할

print("Red\tApple")
```
